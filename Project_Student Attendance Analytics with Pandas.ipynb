## Student Attendance Analytics with Pandas

```python
import pandas as pd
import numpy as np
```

#### Section 0: Data Loading

```python
df = pd.read_excel(r"D:\5th sem\Extra\Student attendance analytics with pandas\aids_attendance.xlsx")
df
```

#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
#-----------------------------Write Solution Code-----------------------------
print(df.head(10)) #10 rows
print(df.shape) #shape
print(df.dtypes) #data types
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
#-----------------------------Write Solution Code-----------------------------
print(df.isnull().sum()) #missing value
print("Columns with null values:",df.isnull().sum()[df.isnull().sum() > 0])
print("count 'never' present in last present",(df["Last Present Date"] == "Never").sum())
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
#-----------------------------Write Solution Code-----------------------------


df.columns = [
    "roll_no", "student_name", "subject", "total_classes",
    "present", "absent", "excused", "attendance_pct",
    "status", "classes_needed_75", "last_present_date",
    "consecutive_absences"
]
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
#-----------------------------Write Solution Code-----------------------------
df["attendance_pct"] = pd.to_numeric(
    df["attendance_pct"].astype("string").str.replace("%", "", regex=False).str.strip(),
    errors="coerce"
)
df["last_present_date"] = pd.to_datetime(
    df["last_present_date"].replace("Never", pd.NaT),
    errors="coerce"
)

print(df[["attendance_pct", "last_present_date"]].dtypes)
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""
#-----------------------------Write Solution Code-----------------------------

print(df[df["status"] == "Critical"].sort_values(by="attendance_pct", ascending=True))
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""
#-----------------------------Write Solution Code-----------------------------

print(df[df["present"] == 0][["roll_no", "student_name", "consecutive_absences"]])
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """

#-----------------------------Write Solution Code-----------------------------
# (a) Safe students using .loc
print("Using loc",df.loc[df["status"] == "Safe"])

# (b) First 15 rows and columns 2-6 using .iloc
print("Using iloc",df.iloc[:15, 1:6])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""
#-----------------------------Write Solution Code-----------------------------

print(df.loc[df["consecutive_absences"] >= 5, ["student_name", "status", "last_present_date"]])
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
#-----------------------------Write Solution Code-----------------------------

print(" Count of students",df.groupby("status")["student_name"].count())
print("\n\nAverage attendance %  ",df.groupby("status")["attendance_pct"].mean())
print("\n\nMaximum consecutive absences",df.groupby("status")["consecutive_absences"].max())
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------

def category(x):
    if x >= 90:
        return "Excellent"
    elif x >= 75:
        return "Good"
    elif x >= 60:
        return "Average"
    else:
        return "Poor"

df["attendance_category"] = df["attendance_pct"].apply(category)

print(df.groupby("attendance_category").size())

```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""
#-----------------------------Write Solution Code-----------------------------

print(df.agg({
    "total_classes": "sum",
    "present": "sum",
    "absent": "sum"
}))
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
#-----------------------------Write Solution Code-----------------------------

print("Group of min, max, and mean of Present",df.groupby("status")["present"].agg(["min", "max", "mean"]))
print("\n\nsum of Absent",df.groupby("status")["absent"].sum())
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
#-----------------------------Write Solution Code-----------------------------

df["risk_score"] = (df["absent"] * 0.4) + (df["consecutive_absences"] * 0.6)

print(df.sort_values("risk_score", ascending=False).head(10))
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
#-----------------------------Write Solution Code-----------------------------

df["classes_needed_numeric"] = df["classes_needed_75"].str.extract(r"(\d+)").fillna(0).astype(int)
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
#-----------------------------Write Solution Code-----------------------------
df["attendance_flag"] = df["attendance_pct"].apply(lambda x: "Defaulter" if x < 75 else "")
df["attendance_flag"] = df["attendance_pct"].apply(lambda x: "Safe" if x >= 75 else "")
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
#-----------------------------Write Solution Code-----------------------------
df["attendance_range"] = pd.cut(
    df["attendance_pct"],
    bins=[0, 50, 75, 90, 100],
    labels=["Critical", "Warning", "Good", "Excellent"],
    include_lowest=True
)
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
#-----------------------------Write Solution Code-----------------------------

print(pd.pivot_table(df, index="status", columns="attendance_category",
                     values="student_name", aggfunc="count", fill_value=0))
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
#-----------------------------Write Solution Code-----------------------------

df["base_name"] = df["student_name"].str.split().str[0].str.upper()
duplicates = df[df.duplicated("base_name", keep=False)]
print(duplicates)
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
#-----------------------------Write Solution Code-----------------------------

Student_Details = pd.DataFrame({
    "roll_no": df["roll_no"],
    "branch": "AI&DS",
    "semester": 5
})

merged_df = pd.merge(df, Student_Details, on="roll_no", how="inner")
print(merged_df)
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""
#-----------------------------Write Solution Code-----------------------------
# Q20 (a) Export all Critical students
critical_students = df[df["status"] == "Critical"]
critical_students.to_csv("critical_students.csv", index=False)
print(critical_students)

# Q20 (b) Export summary statistics grouped by Status
status_summary = df.groupby("status").agg({
    "attendance_pct": "mean",
    "present": "sum",
    "absent": "sum"
})
status_summary.to_csv("status_summary.csv")
print(status_summary)

# Q20 (c) Export Top 10 students by Risk Score
top10_risk = df.sort_values("risk_score", ascending=False).head(10)
top10_risk.to_csv("top10_risk_students.csv", index=False)
print(top10_risk)
```
